### 실습 2. `np.where`로 상태 라벨 만들기
25℃ 이상이면 `high`, 아니면 `normal`인 배열을 만드세요.

```python
# TODO
temp = np.array([18.2, 19.1, 21.0, 24.3, 27.8, 29.1, 25.4, 22.0])
np.where(temp>=25, 'high', 'normal')
```
