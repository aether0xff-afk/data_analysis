## 학습 목표
- 반복 측정 데이터를 축(axis) 관점에서 분석한다.

**핵심 키워드:** `axis`, `mean/std`, `vectorization`, `nan 함수`

```python
import numpy as np
```

```python
# 4개 조건에서 각각 5회 반복 측정
x = np.array([[10.1,10.2,10.0,10.3,10.1],
              [12.0,12.2,11.9,12.1,12.2],
              [14.3,14.1,14.4,14.2,14.3],
              [16.0,16.1,15.9,16.2,16.1]])
print('조건별 평균:', x.mean(axis=1))  # axis 는 따라가라는 뜻이다 axis 1은 열을 따라서 내려 가라는 뜻!
print('반복회차별 평균:', x.mean(axis=0)) # axis 0 은 행을 따라서 오른쪽으로 가는 뜻!
```

### 실습 1. 조건별 변동성
각 행(조건)의 표본표준편차를 구하세요.

```python
# TODO
print('조건(행)별 std: ', x.std(axis=1, ddof=1))   # 표본 이니까 ddof 를 1 로 준다 분모가 N-1 이라는 뜻
```

### 실습 2. 결측값이 있는 배열
NaN을 포함한 배열의 평균을 `np.nanmean`으로 구하세요.

```python
y = x.astype(float).copy() # using x by type=float and copy for not mutantong x
y[1,2] = np.nan # make y's (1,2) value to NaN
print(y)
# TODO
print("y's NaNmean: ", np.nanmean(y, axis = 1)) # 열을 따라서 행별로 !
```
