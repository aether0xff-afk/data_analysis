## 1. NumPy 시작하기

NumPy는 다차원 배열과 수치 연산을 지원하는 파이썬의 핵심 라이브러리이다. `ndarray`의 각 원소는 동일한 `dtype`을 가지며, 일정한 메모리 구조와 내부적으로 최적화된 벡터 연산을 활용하므로 대규모 수치 계산을 효율적으로 처리할 수 있다.

Structured array도 배열의 각 원소는 동일한 structured dtype을 가지지만, 하나의 원소 안에 서로 다른 타입의 필드를 포함할 수 있다.

관례적으로 NumPy를 `np`라는 별칭으로 불러온다. 아래 셀에서 설치된 버전을 확인하시오.

```python
import numpy as np

print(np.__version__)
```

## 2. 배열(array) 만들기

### 2.1 리스트/튜플로부터 만들기
`np.array()`에 리스트나 튜플을 전달하면 ndarray가 생성된다. 아래 코드를 실행하여 결과를 확인하시오.

```python
a = np.array([1, 2, 3, 4, 5])
print(a)

b = np.array((0.1, 0.2, 0.3))   # 튜플도 가능하다.
print(b)
print(b.shape, b.dtype)

# 2차원 배열을 생성한다.
c = np.array([[1, 2, 3], [4, 5, 6]])
print(c)
print(c.shape, c.dtype)
```

### 2.2 dtype 지정하기
생성 시 값으로부터 dtype이 자동으로 추정되지만, `dtype` 인자로 직접 지정할 수도 있다. 다음 예제를 실행하시오.

```python
z = np.array([1, 2, 3], dtype='float64')
print(z, z.dtype)

x16 = np.array([1, 2, 3], dtype=np.int16)
print(x16, x16.dtype)
```

### 2.3 `arange`, `linspace`, `eye`

- `np.arange(start, stop, step)`: 일정한 간격의 값을 갖는 1차원 배열을 생성한다. 정수 인수에서는 파이썬의 `range()`와 유사하지만 결과는 `ndarray`이다.
- `np.linspace(start, stop, num)`: 기본적으로 시작값과 끝값을 모두 포함하여 구간을 균등하게 나눈 `num`개의 값을 생성한다.
- `np.eye(n)`: `n×n` 단위행렬을 생성한다.

실수 간격을 `arange()`에 사용하면 부동소수점 오차로 원소 개수나 끝부분의 값이 예상과 달라질 수 있다. 끝값과 원소 개수가 중요하면 `linspace()`를 사용하는 것이 안전하다.

```python
print(np.arange(10))               # 0 이상 10 미만의 정수
print(np.arange(1, 2, 0.1))        # 실수 간격도 가능하지만 정밀도에 유의한다.
print(np.linspace(0.0, 20.0, 11))  # 양 끝값을 포함한 11개 값
print(np.eye(3))
```

### 2.4 `zeros`, `ones`, `empty`, 난수

- `np.zeros(shape)`: 모든 원소가 0인 배열을 만든다.
- `np.ones(shape)`: 모든 원소가 1인 배열을 만든다.
- `np.empty(shape)`: 원소를 초기화하지 않은 배열을 만든다. 기존 메모리의 임의 값이 보일 수 있으므로 값을 모두 채우기 전에는 연산에 사용하지 않는다.
- `rng.random(shape)`: 반열린 구간 `[0.0, 1.0)`의 난수로 채운 배열을 만든다.

재현 가능한 실습을 위해 권장 난수 생성기인 `np.random.default_rng(seed)`를 사용한다.

```python
print(np.zeros((3, 5)))
print(np.ones((3, 5)))

uninitialized = np.empty((2, 2))
print('초기화하지 않은 배열(실행마다 값이 다를 수 있음):')
print(uninitialized)

rng = np.random.default_rng(42)
print('seed=42로 생성한 난수:')
print(rng.random((3, 2)))
```

### 2.5 크기를 미리 정하지 않고 값 추가하기

빈 배열을 만들고 `np.append()`로 값을 추가할 수 있다. 그러나 `np.append()`는 기존 배열을 제자리에서 늘리지 않고 매번 새 배열을 반환하므로 반복 횟수가 많으면 비효율적이다. 대량의 값을 모을 때는 파이썬 리스트에 먼저 추가한 뒤 마지막에 `np.array()`로 변환하는 방법이 일반적으로 더 효율적이다.

`np.array([])`의 기본 dtype은 `float64`이므로 정수 배열이 필요하면 dtype을 명시한다.

```python
arr = np.array([], dtype=np.int64)
for i in range(3):
    arr = np.append(arr, [1, 2, 3])

print(arr)
print('dtype:', arr.dtype)
```

## 4. `shape`과 `dtype` 변경

### 4.1 `reshape`

전체 원소 개수를 유지한 채 배열의 모양을 바꿀 수 있다. 함수형 `np.reshape(arr, newshape)`과 메서드형 `arr.reshape(newshape)`을 모두 사용할 수 있다.

`np.arange(1, 10)`의 결과는 `(1, 9)`가 아니라 shape이 `(9,)`인 **1차원 배열**이라는 점에 유의한다.

```python
mat = np.arange(1, 10)             # 원소가 9개인 1차원 배열, shape=(9,)
print(mat)
print('변경 전 shape:', mat.shape)

mat2 = mat.reshape((3, 3))         # 3행 3열의 2차원 배열
print(mat2)
print('변경 후 shape:', mat2.shape)

mat3 = np.reshape(mat, (3, 3))     # 함수형으로 동일한 모양 생성
print(mat3)
```

`shape` 속성은 배열의 현재 모양을 확인하는 데 사용한다. 과거에는 `arr.shape = newshape`처럼 직접 대입하기도 했지만, 현재 공식 문서에서는 이 방식을 권장하지 않으며 폐기될 수 있다고 안내한다. 모양을 바꿀 때는 `reshape()`를 사용한다.

```python
x = np.arange(0, 9, 1.0)
print('변경 전:', x.shape)

x = x.reshape(3, 3)
print('변경 후:', x.shape)
print(x)
```

### 4.2 `astype()`으로 dtype 변경하기

`astype()`은 지정한 dtype의 새로운 배열을 반환하며, 원본 배열의 dtype은 바꾸지 않는다. 다음 코드로 원본과 변환 결과를 비교하시오.

```python
a = np.arange(3)
as_float = a.astype(np.float64)
as_int16 = a.astype(np.int16)

print('원본       :', a, a.dtype)
print('float64 변환:', as_float, as_float.dtype)
print('int16 변환  :', as_int16, as_int16.dtype)
```

## 5. 인덱싱과 슬라이싱

### 5.1 기본 인덱싱과 슬라이싱

`[]`와 `:`를 이용해 원소나 부분 배열에 접근한다. 인덱스는 0부터 시작하고 음수 인덱스는 뒤에서부터 센다. 2차원 배열에서 `mat[행 인덱스, 열 인덱스]` 형식으로 접근한다.

```python
data = np.array([1.2, -1.3, 2.2, 5.3, 3.7])
print(data[0])
print(data[0:3])
print(data[-1])
print(data[0:-1])
print([float(data[0]), float(data[2]), float(data[4])])

mat = np.arange(9).reshape(3, 3)
matt = np.linspace(0,8,9).astype(np.int64).reshape(3,3)
print(matt)
print(mat)
print(mat[1, 2])     # 행 인덱스 1, 열 인덱스 2: 두 번째 행, 세 번째 열
print(mat[1:3, :])   # 행 인덱스 1과 2: 두 번째 행부터 세 번째 행까지
```

```python
matt = np.linspace(0,8,9).astype(np.int64).reshape(3,3)
print(matt)

#해랏! 꼭 해보기 인덱싱이랑 리셰이핑이랑 등등 다 해보자! linspace 로도! 할 수 있어!
```

### 5.2 정수 배열을 이용한 고급 인덱싱

정수 배열을 인덱스로 사용하면 해당 위치의 원소만 선택할 수 있다. 이러한 **고급 인덱싱은 원본과 메모리를 공유하지 않는 복사본**을 만든다.

```python
a = np.array([1.2, -1.3, 2.2, 5.3, 3.7])
idx = [0, 3]
selected = a[idx]

print('선택 결과:', selected)
selected[0] = 999
print('선택 결과 수정:', selected)
print('원본은 유지됨 :', a)
```

### 5.3 불린(Boolean) 인덱싱
조건식을 배열에 적용하면 True/False로 이루어진 배열이 생성되며, 이를 인덱스로 사용하여 조건을 만족하는 원소만 추출할 수 있다. 다음 코드를 실행하시오.

```python
x = np.array([1.2, -1.3, 0., 2.2, 0., 5.3, 3.7])
print(x > 2)
print(x[x > 2])
```

### 5.4 `nonzero`, `where`

`np.nonzero(x)`는 0이 아닌 원소의 위치를 **차원별 인덱스 배열의 튜플**로 반환한다. 불린 배열에 적용하면 `True`인 위치를 반환한다.

인수를 하나만 사용한 `np.where(condition)`은 `np.asarray(condition).nonzero()`와 같은 형태의 결과를 반환한다. `np.where(condition, x, y)`처럼 세 인수를 사용하면 조건이 참인 위치에서는 `x`, 거짓인 위치에서는 `y`의 값을 선택한다.

```python
x = np.array([1.2, -1.3, 0.0, 2.2, 0.0, 5.3, 3.7])

print('0이 아닌 위치  :', np.nonzero(x))
print('2보다 큰 위치  :', np.nonzero(x > 2))
print('where 한 인수  :', np.where(x > 2))
print('where 세 인수  :', np.where(x > 2, x, 0.0))
```

## 6. 배열 합치기

- `np.concatenate((A, B, ...), axis=...)`: 지정한 축을 따라 배열을 이어 붙인다.
- `np.vstack((A, B, ...))`: 배열을 세로로 쌓는다.
- `np.hstack((A, B, ...))`: 배열을 가로로 이어 붙인다.
- `np.append(arr, values, axis=None)`: 값을 덧붙인 **새 배열**을 반환한다. `axis`를 생략하면 입력을 평탄화한 뒤 이어 붙인다.

`np.append()`는 파이썬 리스트의 `append()`와 달리 기존 배열을 제자리에서 바꾸지 않는다.

```python
import numpy as np
```

```python
p = np.array([[1, 1], [2, 2]])
q = np.array([[5, 6]])
print(np.concatenate((p, q), axis=0))

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.vstack((a, b)))
print(np.hstack((a, b)))
print(np.append(a, b))
```

## 7. 배열 연산

### 7.1 요소 단위 연산과 제자리 연산

shape이 같은 배열끼리의 사칙연산은 같은 위치의 원소끼리 수행된다. shape이 다르더라도 브로드캐스팅 규칙을 만족하면 연산할 수 있다.

`a += b`는 기존 배열의 메모리에 결과를 저장하는 **제자리 연산**이다. 반면 `a = a + b`는 새 결과 배열을 만든 뒤 변수 `a`가 그 배열을 가리키게 한다. 현재 예제의 계산값은 같지만 객체의 동일성, 공유 관계, dtype 변환 가능 여부가 다를 수 있다.

```python
a = np.array([1, 2, 3, 4])
b = np.array([4, 5, 6, 7])

print(a + b)
print(a * b)
print(a ** 2)
print(a + 2)
print(a < 3)

inplace = np.array([1, 2, 3, 4])
inplace_alias = inplace
inplace += b
print('제자리 연산 결과:', inplace)
print('같은 객체인가?  :', inplace is inplace_alias)

new_result = np.array([1, 2, 3, 4])
previous = new_result
new_result = new_result + b
print('새 배열 연산 결과:', new_result)
print('같은 객체인가?   :', new_result is previous)
```

### 7.2 Universal 함수
`sin`, `cos`처럼 배열 전체에 elementwise로 적용되는 수학 함수를 universal 함수라고 한다.

```python
theta = np.arange(0., 2 * np.pi, 0.5)
print(np.sin(theta))
print(np.cos(theta))
print(np.sqrt(np.array([1, 4, 9, 16])))
```

## 8. 행렬 연산과 선형대수

### 8.1 행렬곱

이 예제처럼 입력이 모두 2차원 배열이면 `np.dot()`, `np.matmul()`, `@`가 같은 행렬곱 결과를 낸다. 그러나 1차원 배열이나 3차원 이상의 배열에서는 `dot`과 `matmul`의 규칙이 서로 다르다. 일반적인 행렬곱의 의도를 명확히 표현할 때는 `@` 또는 `np.matmul()`을 사용하는 것이 좋다.

```python
mat1 = np.array([[1, 2, 3], [3, 4, 5], [5, 6, 7]])
unit_matrix = np.eye(3)

print(np.dot(mat1, unit_matrix))
print(np.matmul(mat1, unit_matrix))
print(mat1 @ unit_matrix)
```

### 8.2 전치행렬
전치행렬은 `.T` 속성 또는 `np.transpose()`로 구할 수 있다.

```python
A = np.arange(9).reshape(3, 3)
print(A.T)
print(np.transpose(A))
```

### 8.3 축(axis) 기준 집계

집계 연산에서 `axis`는 **연산 후 제거되는 축**을 뜻한다.

- `axis=0`: 행 축을 따라 값을 모아 각 열의 결과를 구한다. `(행, 열) → (열,)`
- `axis=1`: 열 축을 따라 값을 모아 각 행의 결과를 구한다. `(행, 열) → (행,)`

“행 방향·열 방향”이라는 표현은 혼동을 일으킬 수 있으므로 “열별 집계·행별 집계”로 결과를 함께 확인한다.

```python
data = np.array([[1, 2, 3], [3, 4, 5]])
print('원본 shape:', data.shape)
print('axis=0, 열별 최댓값:', data.max(axis=0))
print('axis=1, 행별 최댓값:', data.max(axis=1))
print('axis=0, 열별 합계  :', data.sum(axis=0))
```

### 8.4 선형대수 함수
`np.linalg` 모듈에는 고유값 분해, 행렬식, 역행렬, 연립방정식 풀이 등의 함수가 포함되어 있다. 아래 예제를 실행하시오.

```python
M = np.array([[1, 2, -1], [3, 7, 0], [0, 4, -1]])

eigvals, eigvecs = np.linalg.eig(M)
print('고유값 :', eigvals)
print('고유벡터:\n', eigvecs)

print('행렬식 :', np.linalg.det(M))
print('역행렬 :\n', np.linalg.inv(M))

b = np.array([1, 2, 3])
x = np.linalg.solve(M, b)
print('Mx=b 의 해 :', x)
```

## 9. 브로드캐스팅

브로드캐스팅은 shape이 다른 배열을 요소 단위로 연산할 때 NumPy가 두 배열을 호환되는 모양으로 **해석하는 규칙**이다. 작은 배열을 실제로 여러 번 복사하는 것이 아니라 확장된 것처럼 처리하므로 효율적인 경우가 많다.

두 배열의 shape을 오른쪽 차원부터 비교하며, 각 차원이 다음 중 하나를 만족하면 호환된다.

1. 두 차원의 크기가 같다.
2. 두 차원 중 하나의 크기가 1이다.

조건을 만족하지 않으면 `ValueError`가 발생한다. 아래에서 `(3, 3)`, `(3,)`, `(1, 3)`, `(3, 1)` 배열의 결과를 비교하시오.

```python
A = np.arange(9.).reshape(3, 3)
x = np.array([1., 0, 0])        # (3,) 형태의 배열이다.
y = x.reshape(1, 3)             # (1,3) 형태의 배열이다.
z = x.reshape(3, 1)             # (3,1) 형태의 배열이다.

print('A + 스칼라\n', A + 1)
print('A + (3,) 배열\n', A + x)
print('A + (1,3) 배열\n', A + y)
print('A + (3,1) 배열\n', A + z)
print('(1,3) + (3,1)\n', y + z)
```

### 참고: 행렬 연산 함수와 브로드캐스팅
`matmul`, `dot`처럼 선형대수 연산을 수행하는 함수는 1차원 배열을 어떻게 취급하는지 주의해서 확인해야 한다. `A @ x`처럼 2차원 배열과 1차원 배열을 곱하면 결과도 1차원 배열로 반환된다.

```python
A = np.arange(9.).reshape(3, 3)
x = np.array([2, 0, 1])   # (3,) 형태의 1차원 배열이다.
print(np.matmul(A, x))    # (3,3) @ (3,) 연산 결과는 (3,) 이다.
```

## 10. 정렬, 탐색, 이진탐색

- `np.sort(x)`: 정렬된 복사본을 반환한다.
- `np.argsort(x)`: 정렬된 순서를 만드는 인덱스를 반환한다.
- `np.amin(x)`, `np.amax(x)`: 최솟값과 최댓값을 반환한다.
- `np.argmin(x)`, `np.argmax(x)`: 최솟값과 최댓값이 처음 나타난 위치를 반환한다.

“정렬 순서상 두 번째로 큰 원소”와 “서로 다른 값 중 두 번째로 큰 값”은 중복 값이 있을 때 결과가 달라질 수 있다.

```python
x = np.array([9.1, 8.2, 2.3, 3.0, 3.0, 7.6, 5.2])

print('최솟값 :', np.amin(x), '위치:', np.argmin(x))
print('최댓값 :', np.amax(x), '위치:', np.argmax(x))
print('정렬   :', np.sort(x))

order = np.argsort(x)
print('정렬 인덱스:', order)

second_in_order_idx = order[-2]
print('정렬 순서상 두 번째로 큰 원소:', x[second_in_order_idx])

distinct_values = np.unique(x)
if distinct_values.size >= 2:
    print('서로 다른 값 중 두 번째로 큰 값:', distinct_values[-2])
```

### 이진탐색 (`np.searchsorted`)
이미 정렬된 배열에서 특정 값이 들어갈 위치를 빠르게 찾을 때 사용한다. `side='left'`는 C++의 `lower_bound`와, `side='right'`는 `upper_bound`와 동일하게 동작한다. 다음 코드를 실행하여 결과를 확인하시오.

```python
sorted_arr = np.array([0., 13., 26., 30.])

for v in [-1, 0, 5, 13, 15, 30, 35]:
    left = np.searchsorted(sorted_arr, v, side='left')
    right = np.searchsorted(sorted_arr, v, side='right')
    print(f'{v:>4} : left={left}, right={right}')
```

정렬된 배열에 새로운 값들을 정렬 순서를 유지한 채로 삽입할 수도 있다.

```python
base = np.array([1, 1, 2, 4, 7, 7, 11, 13, 13, 13, 15, 20, 25, 26, 27, 30, 45, 70])
new_vals = np.array([5, 7, 9, 45])

positions = np.searchsorted(base, new_vals)
merged = np.insert(base, positions, new_vals)
print(merged)
```

## 11. 배열의 별칭(alias), 뷰(view), 독립 복사(copy)

NumPy 배열의 공유 관계는 다음 세 가지를 구분해야 한다.

1. `y = x`: 복사가 일어나지 않는다. 두 변수는 완전히 같은 배열 객체를 가리키는 **별칭**이다.
2. 기본 슬라이싱 `view = x[1:3]`: 새로운 배열 객체인 **뷰**를 만들지만 원본의 데이터 메모리를 공유한다.
3. `copy = x.copy()`: 데이터 버퍼를 복제하여 독립된 배열을 만든다.

수치 배열에서는 `.copy()`로 만든 배열을 수정해도 원본이 바뀌지 않는다. 다만 object dtype 배열의 경우 배열 안에 저장된 파이썬 객체까지 재귀적으로 복제하는 Python의 `deepcopy`와는 다르므로, `.copy()`를 일반적인 의미의 “깊은 복사”라고 부르지 않는 것이 정확하다.

```python
x = np.array([1, 2, 3])
y = x                    # 복사가 아니라 동일 객체의 별칭
y[0] = 10

print('x:', x)
print('y:', y)
print('y is x:', y is x)
print('메모리 공유:', np.shares_memory(x, y))

z = x.copy()             # 독립된 데이터 복사본
z[0] = -1

print('x:', x)
print('z:', z)
print('z is x:', z is x)
print('메모리 공유:', np.shares_memory(x, z))
```

### 기본 슬라이싱의 뷰와 고급 인덱싱의 복사본

NumPy의 기본 슬라이싱은 일반적으로 원본과 데이터를 공유하는 뷰를 반환한다. 반면 정수 배열이나 불린 배열을 사용하는 고급 인덱싱은 복사본을 반환한다.

파이썬 리스트의 슬라이싱은 새로운 리스트를 만들지만 내부의 중첩 객체는 공유하는 **얕은 복사**이다. 따라서 파이썬 리스트 슬라이싱도 깊은 복사라고 설명해서는 안 된다.

```python
x = np.array([1, 2, 3])

view = x[0:2]            # 기본 슬라이싱: 뷰
view[0] = 100
print('슬라이스 뷰:', view)
print('원본 배열   :', x)
print('메모리 공유 :', np.shares_memory(x, view))

advanced = x[[0, 1]]     # 고급 인덱싱: 복사본
advanced[0] = -999
print('고급 인덱싱 결과:', advanced)
print('원본 배열 유지   :', x)
print('메모리 공유      :', np.shares_memory(x, advanced))

nested = [[1], [2]]
list_slice = nested[:]
list_slice[0].append(99)
print('리스트 원본      :', nested)
print('리스트 슬라이스  :', list_slice)
```

## 12. 데이터 파일 입출력

### 12.1 텍스트 파일: `savetxt` / `loadtxt`

사람이 읽을 수 있는 텍스트 형식으로 배열을 저장하고 불러올 수 있다. 기본 구분자는 공백이며 `delimiter=','`와 같이 지정할 수 있다.

텍스트 파일은 배열의 dtype 정보를 완전하게 보존하지 않는다. `np.loadtxt()`의 기본 dtype은 `float64`이므로 정수로 불러오려면 `dtype=int` 등을 명시한다. 정수를 텍스트로 저장하려면 `fmt='%d'`를 사용할 수 있다.

```python
mat = np.arange(9).reshape(3, 3)
np.savetxt('sample_3x3.txt', mat, fmt='%d')

loaded_default = np.loadtxt('sample_3x3.txt')
loaded_int = np.loadtxt('sample_3x3.txt', dtype=np.int64)

print('기본 dtype으로 읽기:')
print(loaded_default)
print('dtype:', loaded_default.dtype)

print('정수 dtype으로 읽기:')
print(loaded_int)
print('dtype:', loaded_int.dtype)
```

1차원 배열을 `savetxt()`로 저장하면 각 원소가 한 행씩 기록된다. 여러 값으로 이루어진 예제 파일을 다시 읽으면 shape이 `(n,)`인 1차원 배열이 된다. 값이 하나뿐이거나 `ndmin` 등의 옵션을 사용하면 결과 shape이 달라질 수 있으므로 필요한 경우 shape을 확인한다.

```python
v = np.array([1.1, 2.2, 3.1])
np.savetxt('sample_vector.txt', v)

loaded_v = np.loadtxt('sample_vector.txt')
print(loaded_v)
print('shape:', loaded_v.shape)
```

### 12.2 바이너리 파일: `.npy`, `.npz`

- `np.save(file, arr)`: 배열 하나를 `.npy` 형식으로 저장한다.
- `np.savez(file, **kwargs)`: 여러 배열을 압축하지 않은 `.npz` 파일 하나에 저장한다.
- `np.load(file)`: 위 형식의 파일을 읽는다.

`.npy`와 `.npz`는 배열의 shape과 dtype을 함께 보존한다. `np.load()`가 반환한 파일 객체는 `with` 문으로 안전하게 닫는 것이 좋다.

```python
rng = np.random.default_rng(42)
acc = rng.random(10)
dt = 60.0
station = 3

np.savez('sample_acceleration.npz', acc=acc, dt=dt, station=station)

with np.load('sample_acceleration.npz') as loaded:
    print(loaded['acc'])
    print(loaded['dt'])
    print(loaded['station'])
```

### 12.3 `tofile` / `fromfile`

배열을 원시 바이너리 형식으로 빠르게 저장하고 복원할 수 있다. 이 형식에는 dtype, shape, 바이트 순서 등의 메타데이터가 포함되지 않으므로 별도로 관리해야 한다. 복원할 때는 저장한 배열과 같은 dtype을 지정한 뒤 원래 shape으로 변경한다.

```python
arr = np.arange(9.0, dtype=np.float64).reshape(3, 3)
original_shape = arr.shape
original_dtype = arr.dtype

arr.tofile('sample_raw.bin')

restored = np.fromfile('sample_raw.bin', dtype=original_dtype)
restored = restored.reshape(original_shape)
print(restored)
print('dtype:', restored.dtype, 'shape:', restored.shape)
```

## 13. 데이터 타입과 structured array

### 13.1 dtype 추론과 명시

`dtype`을 지정하지 않으면 NumPy가 입력값으로부터 dtype을 추론한다.

- 파이썬 실수 값으로 만든 배열은 일반적으로 `float64`가 된다.
- 파이썬 정수 값으로 만든 배열은 플랫폼의 기본 정수 dtype을 사용한다. 64비트 환경에서는 흔히 `int64`이지만 항상 `int32` 또는 `int64`라고 단정해서는 안 된다.
- 크기를 정확히 고정해야 하는 파일 형식이나 통신 데이터에서는 `np.int32`, `np.float64`처럼 dtype을 명시한다.

```python
default_int = np.array([1, 2, 3])
default_float = np.array([1.0, 2.0, 3.0])
fixed_int16 = np.array([1, 2, 3], dtype=np.int16)
fixed_float32 = np.array([1, 2, 3], dtype=np.float32)

print('추론된 정수 dtype:', default_int.dtype)
print('추론된 실수 dtype:', default_float.dtype)
print('명시한 정수 dtype:', fixed_int16.dtype)
print('명시한 실수 dtype:', fixed_float32.dtype)
print(np.dtype('int32') == np.dtype(np.int32))
```

주로 사용하는 NumPy 데이터 타입은 다음과 같다.

- 불린형: `np.bool_`
- 부호 있는 정수형: `np.int8`, `np.int16`, `np.int32`, `np.int64`
- 부호 없는 정수형: `np.uint8`, `np.uint16`, `np.uint32`, `np.uint64`
- 실수형: `np.float16`, `np.float32`, `np.float64`
- 복소수형: `np.complex64`, `np.complex128`

고정 크기 dtype은 표현 범위를 넘는 값에서 오버플로가 발생할 수 있으므로 `np.iinfo()`와 `np.finfo()`로 범위를 확인할 수 있다.

### 13.2 Structured array

Structured array는 C의 구조체처럼 하나의 원소 안에 이름, 나이, 키 등 여러 필드를 포함할 수 있다. 배열의 모든 원소는 동일한 structured dtype을 가지며, 각 필드는 서로 다른 dtype을 가질 수 있다.

`'S10'`은 바이트 문자열이므로 출력에 `b'Alice'`처럼 `b`가 붙고 한글 처리에도 적합하지 않다. 일반 문자열과 한글을 저장하려면 최대 10개의 유니코드 문자를 저장하는 `'U10'`을 사용한다.

```python
student_dtype = [
    ('name', 'U10'),
    ('age', np.int32),
    ('height', np.float32),
]

students = np.array(
    [('김민준', 19, 173.0), ('이서연', 20, 181.0)],
    dtype=student_dtype,
)

print(students)
print(students['name'])
print(students['age'])
print(students[0])
```

여러 개의 리스트에 흩어져 있는 자료도 같은 길이의 structured array를 만든 뒤 필드별로 대입하여 모을 수 있다.

```python
names = ['김민준', '이서연', '박도윤']
ages = [25, 45, 37]
heights = [173.0, 181.0, 165.0]

student_dtype = [
    ('name', 'U10'),
    ('age', np.int32),
    ('height', np.float32),
]

students2 = np.zeros(3, dtype=student_dtype)
students2['name'] = names
students2['age'] = ages
students2['height'] = heights

print(students2)
```

## 14. 기타: `nan`, `inf` 활용

NumPy는 `nan`(Not a Number)과 양·음의 무한대인 `inf`, `-inf`를 부동소수점 특수값으로 제공한다.

- `nan`은 부동소수점 배열에서 결측 또는 정의되지 않은 값을 나타낼 때 사용할 수 있다. 정수 배열은 `nan`을 직접 저장할 수 없다.
- `np.isnan()`은 각 원소가 `nan`인지 검사한다.
- `np.nanmax()`, `np.nanmean()` 등은 `nan`을 제외하고 계산한다. 단, 대상이 모두 `nan`이면 경고나 `nan` 결과가 나타날 수 있다.
- 최댓값 탐색의 초기값으로 `-np.inf`, 최솟값 탐색의 초기값으로 `np.inf`를 사용할 수 있다.

```python
values = np.array([2.0, 3.0, 1.0, np.nan, 5.0])
print(values)
print('dtype:', values.dtype)
print(np.isnan(values))
print(np.nanmax(values))

maximum = -np.inf
data = [3, 7, 2, 9, 4]
for value in data:
    if value > maximum:
        maximum = value
print('최댓값:', maximum)

minimum = np.inf
for value in data:
    if value < minimum:
        minimum = value
print('최솟값:', minimum)
```
