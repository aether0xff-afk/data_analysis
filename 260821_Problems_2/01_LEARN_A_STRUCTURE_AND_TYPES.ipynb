
## A5. UCI의 `-200`은 숫자가 아니다

즉 `-200`이 float/int로 잘 읽혔다고 해서 실제 측정값으로 사용하면 안 된다.

모든 numeric column에서 `-200`을 `np.nan`으로 바꾸고,
각 열의 결측치 개수를 내림차순으로 출력해라.

힌트:

```python
df[numeric_cols] = df[numeric_cols].replace(-200, np.nan)
df[numeric_cols].isna().sum()
```
